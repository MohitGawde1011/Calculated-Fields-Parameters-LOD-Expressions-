1. What is a Calculated Field and why is it important in business dashboards?

-    A Calculated Field in Tableau is a new field created using a formula based on existing data fields.

Example:
Profit Ratio = SUM([Profit]) / SUM([Sales])

Importance:

Helps create new business metrics.
Allows data analysis without changing the original dataset.
Makes dashboards more useful and interactive.
Helps businesses measure KPIs such as profit margin, growth rate, and sales performance.


2. Why can’t parameters change visuals directly in Tableau?

-   Parameters store a single value such as a number, date, or text. They do not directly control a visual.

They need to be connected to a calculated field, filter, or reference line to affect a visualization.

Example: A parameter can allow a user to choose Sales or Profit, but a calculated field is needed to make the chart switch between them.

3. When should EXCLUDE LOD be used?

-   EXCLUDE LOD is used when you want to calculate a value while ignoring one or more dimensions that are present in the view.

Example:
If a view shows Region and Product, but you want to calculate the average sales for each Region regardless of Product, you can use:

{EXCLUDE [Product] : AVG([Sales])}

Use it when: You want to compare detailed values with a higher-level overall value.

4. Difference between FIXED and INCLUDE LOD expressions

-   | FIXED LOD                                                                            | INCLUDE LOD                                                                                 |
| ------------------------------------------------------------------------------------ | ------------------------------------------------------------------------------------------- |
| Calculates a value based on the dimensions specifically mentioned in the expression. | Calculates a value by adding the specified dimension to the view's level of detail.         |
| Can ignore dimensions in the view.                                                   | Works with the dimensions already present in the view.                                      |
| Example: `{FIXED [Region] : SUM([Sales])}`                                           | Example: `{INCLUDE [Customer] : SUM([Sales])}`                                              |
| Useful for calculating fixed-level metrics such as regional sales.                   | Useful for adding more detailed calculations, such as customer-level sales within a region. |


Task 1: Delivery Performance Classification — Calculated Field

Business Need: Classify orders based on delivery time.

Steps:

Open Tableau and connect the Blinkit Orders Dataset.
Go to Data Pane → Create Calculated Field.
Name it Delivery Performance.
Enter:

IF [Delivery_Time_Min] <= 30 THEN
    "Fast"
ELSE
    "Delayed"
END

Result:

Delivery time ≤ 30 minutes → Fast
Delivery time > 30 minutes → Delayed

You can use this field in a bar chart to compare Fast vs. Delayed orders.

2: Loss-Making Orders Identification

Business Need: Identify orders where the company is making a loss.

Create a calculated field called Loss-Making Order.

IF [Profit] < 0 THEN
    "Loss"
ELSE
    "Profit"
END

Result:

Profit < 0 → Loss
Profit ≥ 0 → Profit

To identify the actual loss-making orders, place Order_ID, Sales, Cost, and Profit in the view and filter Loss-Making Order = Loss.

Task 3: Top N Products Analysis — Parameter

Business Need: Allow management to dynamically select the number of top products.

Step 1: Create the parameter

Go to:

Data Pane → Create Parameter

Set:

Name: Top N Products
Data Type: Integer
Current Value: 10
Allowable Values: Range
Minimum: 1
Maximum: 50
Step Size: 1
Step 2: Create a ranking calculation

Create a calculated field called Product Rank:

RANK(SUM([Sales]), 'desc')
Step 3: Create the Top N filter

Create another calculated field called Top N Product Filter:

RANK(SUM([Sales]), 'desc') <= [Top N Products]

Drag Top N Product Filter to the Filters shelf and select True.

Now, when you change the Top N Products parameter from 5 to 10 to 20, Tableau dynamically displays the selected number of top-performing products.

Alternative: You can also use Tableau's built-in Top → By Field → Top N → SUM(Sales) option and connect N to the parameter.

Task 4: City-Level Benchmarking — FIXED LOD

Business Need: Calculate the average sales for each city independently of category-level filtering.

Create a calculated field called City Average Sales:

{ FIXED [City] : AVG([Sales]) }

This calculates the average sales for each city.

For example:

| City   |      City Average Sales |
| ------ | ----------------------: |
| Mumbai | Average Sales of Mumbai |
| Pune   |   Average Sales of Pune |
| Delhi  |  Average Sales of Delhi |


Why FIXED LOD?

FIXED [City] calculates the value at the City level, regardless of which dimensions are displayed in the view.

Task 5: Product-Level Detail Impact — INCLUDE LOD

Business Need: Calculate average sales at the Product level, while allowing the result to be viewed at a higher level such as Category.

Create a calculated field called Product Average Sales:

{ INCLUDE [Product_Name] : AVG([Sales]) }

This tells Tableau:

"Include Product_Name in the calculation and calculate the average Sales at the product level."

You can then place Category in the view and use Product Average Sales to analyze the product-level sales contribution within each category.
